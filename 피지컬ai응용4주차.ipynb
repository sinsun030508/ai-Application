```python
#[비지도 학습 중 clustering 기본 코드]

from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

# 1. 데이터 준비 (정답 라벨 y는 사용하지 않음)
X, _ = load_iris(return_X_y=True)

# 2. 모델 생성 및 학습 (X만 전달)
model = KMeans(n_clusters=3, random_state=42)
model.fit(X)

# 3. 예측 (각 데이터가 할당된 임의의 군집 번호 출력)
cluster_labels = model.predict(X[:5])

print("[군집] 할당된 클러스터 번호:", cluster_labels)



```

```python
#[비지도 학습 중 Dimensionality Reduction 기본 코드]

from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# 1. 데이터 준비 (정답 라벨 y는 사용하지 않음)
X, _ = load_iris(return_X_y=True)

# 2. 모델 생성 및 변환 (4차원 -> 2차원으로 축소)
pca = PCA(n_components=2)
X_reduced = pca.fit_transform(X)

# 3. 크기 변화 확인
print("[차원축소] 원본 데이터 크기 (샘플수, 특성수):", X.shape)
print("[차원축소] 압축 데이터 크기 (샘플수, 특성수):", X_reduced.shape)
print("[차원축소] 압축된 첫 번째 샘플:", X_reduced[0])

```
