while 문을 이용하여 아래 내용을 구현하시오

-손실(loss)가 2.0에서 시작해 매 epoch마다 0.8배로 줄어듭니다


-손실이 0.5 이하가 되려면 몇 epoch가 걸리는지 while 문으로 구하고, 그 때의 손실도 출력하세요

```python
loss = 2.0
epoch = 0

while loss > 0.5:
    loss = loss * 0.8
    epoch += 1
print("걸린 epoch : ", epoch)
print("걸린 최종 loss : ", loss)
```

리스트를 받아 가장 작은 값은 0, 가장 큰 값은 1이 되도록 바꿔주는 함수 normalize를 만드세요

(numpy X이고 순수 파이썬으로 작성) (hint) 공식은 (v-최소값)/(최대값 - 최솟값), 최소최대는 min(), max()임




```python
data = [10,20,30,40,50]

def normalize(data):
    min_val = min(data)
    max_val = max(data)
    return [(v - min_val) / (max_val - min_val) for v in data]

print(normalize(data))
print(normalize([3,1,4,1,5]))
```

딕셔너리 모델이 내놓은 예측이름들을 세어서, 많이 나온 순서대로 출력하세요





```python
pred = ["dog", "cat", "dog", "bird", "dog", "cat"]

count = {}
for p in pred:
  count[p] = count.get(p,0) +1

ranked = sorted(count.items(), key=lambda kv: kv[1], reverse=True)

print(count)
print(ranked)
```

정답과 예측을 비교해 틀린 위치를 모두 찾고 틀린 개수를 출력하시오

```python
y_true = [0,1,1,0,1,0,1,1]
y_pred = [0,1,0,0,0,0,1,1]

wrong = []

for i, (t,p) in enumerate(zip(y_true, y_pred)):
  if t != p:
    wrong.append(i)

print(wrong)
print(len(wrong))
print((len(y_true) - len(wrong)/len(y_true)))
```

이중 for문


- 2차원 리스트에서각 행의 합과 전체합을 이중 반복문으로 구하기


```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]

row_sum = []
total_sum = 0

for row in matrix:
  row_sum.append(sum(row))
  total_sum += sum(row)

print("행별 합:", row_sum)
print("전체합 :", total_sum)
```
